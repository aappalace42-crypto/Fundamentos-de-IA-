```python
# Grafos dirigidos y ponderados
import networkx as nx
import matplotlib.pyplot as plt

# Creamos el grafo
grafo = nx.DiGraph()
grafo.add_weighted_edges_from([
    ("C", "F", 5),
    ("F", "G", 4),
    ("C", "D", 2),
    ("D", "G", 8),
    ("E", "G", 3),
    ("D", "E", 2),
    ("C", "A", 7),
    ("B", "A", 4),
    ("B", "D", 3),
    ("B", "E", 6)
])

# imprimimos el grafo
pos = {
    "G": (7.7, 5.5),
    "F": (4.0, 4.0),
    "C": (3.3, 2.6),
    "D": (5.4, 2.6),
    "E": (7.2, 2.1),
    "A": (3.3, 0.0),
    "B": (5.4, 0.0),
}

nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)

labels = nx.get_edge_attributes(grafo, "weight")

nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)

plt.show()
```

```python
# Calculamos el costo 
def costo_ruta(grafo2, ruta):
    costo = 0
    for i in range(len(ruta)-1):
        costo += grafo2[ruta[i]][ruta[i+1]]["weight"]
    return costo
```

```python
# Imprimimos el costo
rutas = [
    ["C", "F", "G"],
    ["C", "D", "G"],
    ["C", "D", "E", "G"],
    ["C", "A"],
]
for i, ruta in enumerate(rutas, start=1):
    costo = costo_ruta(grafo, ruta)
    print(f"Ruta {i}: {ruta} -> {costo} unidades")
```

```python
#Elegir la ruta más corta
costos = [costo_ruta(grafo, ruta) for ruta in rutas]
mejor_ruta = costos.index(min(costos))
```

```python
#Imprimir la mejor ruta
print(f"La mejor ruta es:{mejor_ruta+1}")
print(f"El camino más corto es: {costos[mejor_ruta]}")
```
