```python
# Grafos dirigidos y ponderados 
import networkx as nx
import matplotlib.pyplot as plt
```

```python
#Creamos el grafo 
grafo=nx.DiGraph()
grafo.add_weighted_edges_from([
    ("A","B",5),
    ("A","C",2),
    ("B","D",4),
    ("C","E",5),
    ("C","F",6)
])
```

```python
# imprimimos el grafo
pos = nx.spring_layout(grafo, seed=10)

nx.draw(
    grafo,
    pos,
    with_labels = True,
    node_size = 1500,
    arrows = True
)

labels = nx.get_edge_attributes(grafo, "weight")

nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels = labels
)

plt.show()
```

```python
# Mostramos las rutas

rutas = list(
    nx.all_simple_paths(
        grafo,
        source = "A",
        target = "F"
    )
)

# Imprimimos las rutas

for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```
