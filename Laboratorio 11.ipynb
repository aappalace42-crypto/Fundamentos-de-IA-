```python
import networkx as nx
import matplotlib.pyplot as plt
import networkx as nx


grafo = nx.DiGraph()
grafo.add_weighted_edges_from([
    ("A","B",8), ("A","E",4), ("A","D",5),
    ("B","C",3), ("B","F",4),
    ("C","G",2), ("C","F",9),
    ("D","E",3), ("D","H",6),
    ("E","I",5),
    ("F","E",3), ("F","K",8),
    ("G","K",3), ("G","L",2),
    ("H","I",4), ("H","M",2),
    ("I","J",4), ("I","M",1), ("I","N",3),
    ("J","O",9),
    ("K","J",6), ("K","P",1),
    ("L","K",5), ("L","P",3),
    ("M","N",1),
    ("N","O",2),
    ("O","P",12),
])
```

```python
def costo_ruta(grafo2, ruta):
    costo = 0
    for i in range(len(ruta)-1):
        costo += grafo2[ruta[i]][ruta[i+1]]["weight"]
    return costo
```

```python
ruta_optima = nx.dijkstra_path(grafo, "A", "P", weight="weight")
costo_optimo = nx.dijkstra_path_length(grafo, "A", "P", weight="weight")

print(f"Ruta óptima: {ruta_optima}")
print(f"Costo total: {costo_optimo}")
```

```python
# Todas las rutas simples posibles entre origen y destino
todas_las_rutas = list(nx.all_simple_paths(grafo, source="A", target="P"))

# Calculamos el costo de cada una reutilizando tu función
costos = [costo_ruta(grafo, ruta) for ruta in todas_las_rutas]

# Índices de la más corta y la más larga
i_corta = costos.index(min(costos))
i_larga = costos.index(max(costos))

print(f"Camino más corto: {todas_las_rutas[i_corta]} -> {costos[i_corta]} unidades")
print(f"Camino más largo: {todas_las_rutas[i_larga]} -> {costos[i_larga]} unidades")
```
