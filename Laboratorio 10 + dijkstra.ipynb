```python
#algoritmo dijkstra 
# encuentren el camino mas corto de un nodo
'hasta el destino en un grupo de con pesos no negativos'
import networkx as nx
import matplotlib.pyplot as plt
```

```python
grafo = nx.DiGraph()
grafo.add_weighted_edges_from([
    ("Centro", "Norte", 10),
    ("Centro", "Sur", 15),
    ("Centro", "Oeste", 25),
    ("Norte", "Este", 8),
    ("Norte", "Sur", 30),
    ("Sur", "Oeste", 9),
    ("Oeste", "Este", 15),
    ("Este", "Aeropuerto", 20),
    ("Oeste", "Aeropuerto", 18),
])

pos = {
    "Aeropuerto": (2.0, 5.0),
    "Oeste": (5.5, 4.0),
    "Este": (3.0, 3.0),
    "Norte": (1.5, 2.0),
    "Sur": (3.5, 0.5),
    "Centro": (6.0, 1.5),
}

nx.draw(grafo, pos, with_labels=True, node_size=1500, arrows=True)
labels = nx.get_edge_attributes(grafo, "weight")
nx.draw_networkx_edge_labels(grafo, pos, edge_labels=labels)
plt.show()
```

```python
# Utilizamos algoritmo para obtener la ruta más corta
ruta_mas_corta = nx.dijkstra_path(grafo, source="Centro", target="Aeropuerto", 
                                    weight='weight')

tiempo_total = nx.dijkstra_path_length(grafo, source="Centro", target="Aeropuerto", 
                                         weight="weight")
```

```python
tiempo_total = nx.dijkstra_path_length(grafo, source="Centro", 
                                         weight="weight")

# Imprimir
print(f"La ruta más corta es: {ruta_mas_corta}")
print(f"El tiempo total es: {tiempo_total} minutos")
```
