```python
import matplotlib.pyplot as plt
import networkx as nx
```

```python
grafoBeach = nx.Graph()
```

```python
grafoBeach.add_weighted_edges_from(
    [
        ("B", "D", 15),
        ("D", "C", 10),
        ("C", "E", 25),
        ("B", "A", 20),
        ("A", "C", 10)
    ]
)
```

```python
pos = nx.spring_layout (
    grafoBeach,
    seed=10
)
nx.draw (
    grafoBeach,
    pos,
    with_labels=True,
    node_size=1500
    )
labels=nx.get_edge_attributes(
    grafoBeach,
    "weight"
)

nx.draw_networkx_edge_labels(
    grafoBeach,
    pos,
    edge_labels=labels
)

plt.show()

```

```python
# --- Resolver: ruta más corta de B (Veracruz) a E (Playa del Carmen) ---
ruta = nx.shortest_path(grafoBeach, source="B", target="E", weight="weight")
distancia = nx.shortest_path_length(grafoBeach, source="B", target="E", weight="weight")

print("Ruta más corta:", " → ".join(ruta))
print("Distancia total:", distancia)
```

```python
# Mostramos rutas y pesos
rutas = list(
    nx.all_simple_paths(
        grafoBeach,
        source = "B",
        target = "E"
    )
)

def costo_ruta(grafoBeach, ruta):
    costo = 0
    for i in range(len(ruta)-1):
        costo += grafoBeach[ruta[i]][ruta[i+1]]["weight"]
    return costo

for i, ruta in enumerate (rutas,start=1):
    costo=costo_ruta(grafoBeach, ruta)
    print (f"Ruta{i}={costo} minutos")
```
