```python
###LABORATORIO 3 PREPOSICIONES 

### objetivo: ** Un sistema inteligente supervisa una red empresarial y debe determinar si existe una condición crítica de servicio
```

De los datos que recibe el sistema:

P: El servidor principal está disponible
Q: La latencia supera el límite establecido
R: La pérdida de paquetes supera el límite permitido
S: El tráfico de red es anormalmente alto
T: El servidor de respaldo está disponible
U: El sistema de monitoreo está activo


#Reglas 
Según la diapositiva, existe condición crítica cuando:

el servidor principal no está disponible y
existe latencia alta y
existe pérdida de paquetes alta y, 
el tráfico es anormalmente alto y, 
el servidor de respaldo No está disponible y ,  
el sistema de monitoreo está activo

```python
# Definición de variables (valores de prueba)
servidorPrincipal = True    # P: el servidor principal está disponible
latenciaAlta      = True    # Q: la latencia supera el límite
perdidaAlta       = True    # R: la pérdida de paquetes supera el límite
traficoAlto       = True    # S: el tráfico es anormalmente alto
servidorRespaldo  = False   # T: el servidor de respaldo está disponible
monitoreoActivo   = True    # U: el sistema de monitoreo está activo

condicionCritica = (servidorPrincipal == True and latenciaAlta == True and
                     perdidaAlta == True and traficoAlto == True and
                     servidorRespaldo == False and monitoreoActivo == True)

print("¿Existe una condición crítica?", condicionCritica)
```

```python
print("== Sistema de Monitoreo de Red ==")
print("Development by Aalejandro")

servidorPrincipal = input("¿El servidor principal está disponible? (s/n): ").lower() == "s"
latenciaAlta      = input("¿La latencia supera el límite establecido? (s/n): ").lower() == "s"
perdidaAlta       = input("¿La pérdida de paquetes supera el límite permitido? (s/n): ").lower() == "s"
traficoAlto       = input("¿El tráfico de red es anormalmente alto? (s/n): ").lower() == "s"
servidorRespaldo  = input("¿El servidor de respaldo está disponible? (s/n): ").lower() == "s"
monitoreoActivo   = input("¿El sistema de monitoreo está activo? (s/n): ").lower() == "s"

condicionCritica = (servidorPrincipal == False and latenciaAlta == True and
                     perdidaAlta == True and traficoAlto == True and
                     servidorRespaldo == False and monitoreoActivo == True)

print("Decisión del sistema:")
if condicionCritica:
    print(" ALERTA: Existe una condición crítica de servicio!!!")
else:
    print("Todo en orden, no hay condición crítica.")
```
