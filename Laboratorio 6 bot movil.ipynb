#Descripcion: encuentra la ruta 
#Situacion: un robot móvil debe desplazarse dentro de una planta industrial desde una estacion de carga hasta una estacion de trabajo desde el punto A al punto J 


```python
import matplotlib.pyplot as plt  # Librerias para graficar
import networkx as nx            # librerias que me permite trabajar con estructuras complejas como grafos
```

```python
# Creamos una instancia con la clase graph
# graph es una coleccion de nodos
grafo = nx.Graph()
```

```python
# Creamos el grafo utilizando el metodo add_edges_from
grafo.add_edges_from(
    [
        ("A", "B"),
        ("A", "C"),
        ("B", "D"),
        ("D", "F"),
        ("F", "E"),
        ("F", "H"),
        ("E", "G"),
        ("C", "G"),
        ("H", "I"),
        ("I", "J"),
        ("G", "J"),
    ]
)
```

```python
# mostrar el grafo
nx.draw(
    grafo,
    with_labels=True,
    node_size=1000,
    node_color="#F54927",
    node_shape="o",
)
plt.show()
```

```python
# Mostrar las rutas posibles del grafo
rutas = list(
    nx.all_simple_paths(grafo, source="A", target="J")
)
```

```python
# Imprimir las rutas
print("Rutas encontradas: ")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
# Mostrar el total de pasos (costo) de cada ruta
pasos = []
for i, ruta in enumerate(rutas, start=1):
    p = len(ruta) - 1
    pasos.append(p)
    print(f"pasos de la ruta {i}: ", p)
```

```python
# Aplicamos la logica proposicional para determinar la mejor ruta
costo_minimo = min(pasos)
```

```python
# Proposicion Pi: la ruta i es la de menor costo
mejores = []
for i, p in enumerate(pasos, start=1):
    Pi = (p == costo_minimo)
    print(f"P{i} = {Pi}   (Ruta {i} tiene costo {p})")
    if Pi:
        mejores.append(i)
```

```python
print()
if len(mejores) == 1:
    idx = mejores[0]
    print(f"La ruta {idx} es la mejor opcion.")
    print(f"Ruta: {rutas[idx - 1]}")
    print(f"Costo: {costo_minimo} pasos")
else:
    print("Hay un empate entre las rutas:", mejores)
    print("Todas tienen el costo minimo de", costo_minimo, "pasos, elige la que prefieras.")

# Relacion con la busqueda en IA:
# El robot representa un agente inteligente que debe navegar en un entorno modelado como grafo.
# Encontrar todas las rutas posibles (all_simple_paths) equivale a la etapa de
# exploracion del espacio de estados, y elegir la ruta de menor costo equivale a un
# algoritmo de busqueda no informada de costo uniforme (como BFS cuando todas las
# aristas tienen el mismo costo).
```
