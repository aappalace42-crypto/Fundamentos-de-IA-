#Problema: Realizar un sistema que identifique 
si una planta necesita riego para activar el sensor 
#Parametros
#Si el nivel de humedad es menor al 40% la planta necesita riego 
#SI el nivel de humedad es igual o mayor al 40% , la planta no necesita riego

```python
print("Sistema de riego inteligente")

planta = input("Escriba el nombre de la planta: ")
humedad = float(input("Escriba el porcentaje de humedad: "))

print("Datos Almacenados")
print("Planta: ", planta)
print("Humedad: ", humedad)

print("Estado")

if humedad < 40:
    print("La planta necesita riego, sensor activado")
else:
    print("La planta no necesita riego, sensor desactivado")
```
